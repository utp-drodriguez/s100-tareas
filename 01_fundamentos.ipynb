## 1. Variables y tipos de datos

Una **variable** guarda un valor y le da un nombre. En Python no hace falta declarar el tipo: se deduce del valor.

Los tipos básicos son:
- `str` — texto (cadenas)
- `int` — números enteros
- `float` — números decimales
- `bool` — verdadero/falso (`True` / `False`)

```python
# Creamos algunas variables
curso = "Ciencia de Datos"   # str  (texto)
creditos = 3                 # int  (entero)
promedio = 4.5               # float (decimal)
aprobado = True              # bool  (booleano)

# print() muestra un valor en pantalla
print(curso)
print(creditos)

# type() nos dice el tipo de un valor
print(type(curso))
print(type(promedio))
```

## 2. Operaciones básicas

Python funciona como una calculadora. También compara valores.

```python
a = 10
b = 3

print(a + b)   # suma
print(a - b)   # resta
print(a * b)   # multiplicación
print(a / b)   # división (da decimal)
print(a // b)  # división entera
print(a % b)   # residuo (módulo)
print(a ** b)  # potencia (a elevado a b)
```

```python
# Comparaciones: devuelven True o False
print(a > b)    # ¿a es mayor que b?
print(a == b)   # ¿son iguales?  (ojo: == compara, = asigna)
print(a != b)   # ¿son diferentes?
```

## 3. Cadenas de texto (strings)

Las cadenas se escriben entre comillas. Se pueden unir, medir y transformar.

```python
nombre = "Ana"
apellido = "Pérez"

# Unir cadenas (concatenar)
completo = nombre + " " + apellido
print(completo)

# f-strings: insertar variables dentro del texto con f"..."
edad = 28
print(f"{nombre} tiene {edad} años.")

# Algunos métodos útiles
print(completo.upper())        # MAYÚSCULAS
print(completo.lower())        # minúsculas
print(len(completo))           # cantidad de caracteres
print(completo.replace("Ana", "Luisa"))
```

## 4. Listas

Una **lista** guarda varios elementos en orden. Los índices empiezan en **0**.

```python
temas = ["Python", "Estadística", "Machine Learning", "Visualización"]

print(temas)
print(temas[0])     # primer elemento
print(temas[1])     # segundo elemento
print(temas[-1])    # último elemento
print(len(temas))   # cantidad de elementos

# Rebanado (slicing): del índice 0 al 2 (el 2 no se incluye)
print(temas[0:2])

# Agregar un elemento al final
temas.append("Ética")
print(temas)
```

```python
# Recorrer una lista con un bucle for
for tema in temas:
    print("Tema:", tema)
```

## 5. Diccionarios

Un **diccionario** guarda pares `clave: valor`. Es ideal para describir "algo" con varias características.

```python
estudiante = {
    "nombre": "Ana Pérez",
    "edad": 28,
    "grupo": 1,
    "aprobado": True
}

print(estudiante["nombre"])   # acceder por su clave
print(estudiante["edad"])

# Agregar o cambiar un valor
estudiante["correo"] = "ana.perez@utp.ac.pa"
print(estudiante)

# Ver todas las claves y valores
print(estudiante.keys())
print(estudiante.values())
```

## 6. Condicionales (if / elif / else)

Permiten que el programa **tome decisiones**. Fíjate en los dos puntos `:` y en la **sangría** (indentación).

```python
nota = 75

if nota >= 91:
    print("A")
elif nota >= 81:
    print("B")
elif nota >= 71:
    print("C")
else:
    print("Reprobado")
```

## 7. Bucles (for y while)

Los bucles **repiten** acciones. `for` recorre una secuencia; `while` repite mientras se cumpla una condición.

```python
# for con range: números del 1 al 5
for i in range(1, 6):
    print("Número:", i)
```

```python
# while: repetir mientras la condición sea verdadera
contador = 0
while contador < 3:
    print("Vuelta", contador)
    contador = contador + 1
```

## 8. Funciones

Una **función** agrupa código reutilizable. Se define con `def`, recibe **parámetros** y puede **devolver** (`return`) un resultado.

```python
def saludar(nombre):
    return f"¡Hola, {nombre}!"

# Llamar (usar) la función
print(saludar("Ana"))
print(saludar("Carlos"))
```

```python
def promedio(a, b, c):
    return (a + b + c) / 3

resultado = promedio(80, 90, 100)
print("El promedio es:", resultado)
```

## 9. (Bonus) Un vistazo a los datos con pandas

`pandas` es la librería que usaremos para trabajar con datos en forma de **tabla** (filas y columnas). Aquí solo un adelanto — lo veremos a fondo en la próxima clase.

```python
import pandas as pd

# Creamos una tabla pequeña a partir de un diccionario
datos = {
    "nombre": ["Ana", "Luis", "María", "Carlos"],
    "edad":   [28, 34, 25, 41],
    "grupo":  [1, 2, 1, 2]
}

df = pd.DataFrame(datos)

# Ver la tabla
df
```

```python
# Primeras filas
print(df.head())

# Resumen estadístico de las columnas numéricas
print(df.describe())

# Filtrar: personas del grupo 1
print(df[df["grupo"] == 1])
```
